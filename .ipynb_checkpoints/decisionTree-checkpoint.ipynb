```python
# Decision Tree Classification
# Student Academic Performance Prediction
#
# This implementation follows proper machine learning practice:
# - No LabelEncoder for target variable
# - Split data BEFORE preprocessing
# - Pipeline + ColumnTransformer
# - GridSearchCV
# - StratifiedKFold
# - No data leakage

import pandas as pd

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    GridSearchCV
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import OneHotEncoder

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    precision_recall_fscore_support
)


# Load Dataset

df = pd.read_csv("StudentPerformanceFactors.csv")



# Create Target Variable

def performance(score):

    if score < 65:
        return "Low"

    elif score < 70:
        return "Medium"

    else:
        return "High"


df["Performance"] = df["Exam_Score"].apply(performance)


# Features and Target


X = df.drop(columns=["Exam_Score", "Performance"])

y = df["Performance"]


# Explicit class order
class_order = ["Low", "Medium", "High"]



# Split Dataset
# IMPORTANT:
# Split BEFORE preprocessing to avoid data leakage

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)


# Identify Numerical and Categorical Columns

numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "string"]
).columns.tolist()



# Numerical Pipeline

numeric_transformer = Pipeline(

    steps=[

        (
            "imputer",
            SimpleImputer(strategy="median")
        )

    ]

)


# Categorical Pipeline


categorical_transformer = Pipeline(

    steps=[

        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),

        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )

    ]

)


# Combine Both Pipelines

preprocessor = ColumnTransformer(

    transformers=[

        (
            "num",
            numeric_transformer,
            numeric_features
        ),

        (
            "cat",
            categorical_transformer,
            categorical_features
        )

    ]

)



# Build Pipeline

pipeline = Pipeline(

    steps=[

        (
            "preprocessor",
            preprocessor
        ),

        (
            "classifier",
            DecisionTreeClassifier(
                random_state=42
            )
        )

    ]

)



# Hyperparameter Grid
# Small grid suitable for student assignment

param_grid = {

    "classifier__criterion": [
        "gini",
        "entropy"
    ],

    "classifier__max_depth": [
        None,
        5,
        10,
        15,
        20
    ],

    "classifier__min_samples_split": [
        2,
        5,
        10
    ],

    "classifier__min_samples_leaf": [
        1,
        2,
        4
    ]

}


# Stratified Cross Validation

cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=42

)



# Grid Search
# Model selection uses ONLY training data

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring="f1_macro",
    cv=cv,
    n_jobs=-1,
    refit=True,
    return_train_score=True
)


# Train Model Using Grid Search

grid_search.fit(X_train, y_train)

# Best model selected from cross-validation
best_model = grid_search.best_estimator_


# Predict on Test Set

y_pred = best_model.predict(X_test)


# Calculate Performance Metrics

accuracy = accuracy_score(y_test, y_pred)

balanced_acc = balanced_accuracy_score(y_test, y_pred)

macro_precision, macro_recall, macro_f1, _ = precision_recall_fscore_support(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

weighted_precision, weighted_recall, weighted_f1, _ = precision_recall_fscore_support(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)



# Print Results

print("=" * 60)
print("Decision Tree Classification")
print("=" * 60)

print("\nBest Hyperparameters")
print(grid_search.best_params_)

print("\nBest Cross Validation Macro F1")
print(f"{grid_search.best_score_ * 100:.2f}%")

print("\nTest Accuracy")
print(f"{accuracy*100:.2f}%")

print("\nBalanced Accuracy")
print(f"{balanced_acc * 100:.2f}%")

print("\nClassification Report")
print(
    classification_report(
    y_test,
    y_pred,
    labels=class_order,
    target_names=class_order,
    digits=4,
    zero_division=0
) )

print("\nMacro Average")

print(f"Precision : {macro_precision*100:.2f}%")
print(f"Recall    : {macro_recall*100:.2f}%")
print(f"F1 Score  : {macro_f1*100:.2f}%")

print("\nWeighted Average")

print(f"Precision : {weighted_precision*100:.2f}%")
print(f"Recall    : {weighted_recall*100:.2f}%")
print(f"F1 Score  : {weighted_f1*100:.2f}%")

# Confusion Matrix

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=class_order
)

cm_df = pd.DataFrame(
    cm,
    index=class_order,
    columns=class_order
)

print("\nConfusion Matrix")
print("(Actual = Rows, Predicted = Columns)")
print(cm_df)

print("\nProgram Finished Successfully.")
```
